```python
import os 
import zipfile
import glob
```

```python
inv_version = 'v2.5'
```

```python
# select multiple filesn and zip them into a folder
def make_zip(zip_filename, file_list):
    with zipfile.ZipFile(zip_filename,'w',
                         compression=zipfile.ZIP_DEFLATED,
                         compresslevel=9) as zipf:
        for file in file_list:
            zipf.write(file, os.path.basename(file))
```

```python
for year in ['2019', '2020', '2021', '2022', '2023', '2024']:
    
    output_dir = os.path.join('../output', 'ICOS CP Upload')
    fname = f'Munich_Inventory_GHGco_{inv_version}_{year}.zip'
    zip_dir = os.path.join(output_dir, fname)

    # get all files in year folder
    flist = [f for f in glob.glob(os.path.join('../output', year, '*')) if os.path.isfile(f)]

    # exclude hidden files from flist (e.g., .DS_Store)
    flist = [f for f in flist if not os.path.basename(f).startswith('.')]

    print(f'Creating zip file: {zip_dir} with {len(flist)} files.')
    [print(f) for f in flist]

    make_zip(zip_dir, flist)
```
